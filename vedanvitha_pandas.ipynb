##### Q4.  Group-wise Analysis

````python
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'HR', 'IT', 'HR'],
    'Salary': [40000, 30000, 60000, 50000]
})

print(df.groupby('Department')['Salary'].mean())

#expected output
#Department
#HR    40000.0
#IT    50000.0
#Name: Salary, dtype: float64
````

Explanation: groupby('Department') divides the rows into HR and IT groups, and mean() calculates the average salary separately for each department.

For HR:

(30000 + 50000) / 2 = 40000

For IT:

(40000 + 60000) / 2 = 50000

Q1. A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing. 

1.Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().

````python
import pandas as pd

# Load the CSV file
df = pd.read_csv("student_assessment_dirty.csv")

# Display the first 5 rows
print("First 5 rows:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
````

2.Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

````python

df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .str.replace(".", "", regex=False)
)

print(df["Department"])
````

Explanation: We convert all department names to uppercase and remove dots so that different spellings are represented consistently.

3.Identify and remove exact duplicate student assessment records.

````python
# Identify exact duplicate records
print("Duplicate records:")
print(df[df.duplicated()])

# Remove exact duplicates
df = df.drop_duplicates()

print("Number of records after removing duplicates:", len(df))
````

Explanation: duplicated() identifies exact duplicate rows, and drop_duplicates() removes them. In this dataset, the S102 – Ravi record appears twice, so 1 duplicate record is removed

4.Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.

````python
# Handle special values in Marks
df["Marks"] = df["Marks"].replace({
    "Absent": None,
    "NA": None,
    "": None
})

# Extract numeric values and convert to numeric type
df["Marks"] = (
    df["Marks"]
    .astype("string")
    .str.extract(r"(\d+(?:\.\d+)?)", expand=False)
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df[["Student_ID", "Marks"]])
````

Explanation: The Marks column is cleaned by extracting numeric values and converting them to a numeric data type, while absent, NA, and blank values are treated as missing

5.Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.

````python
# Identify students with missing marks
missing_marks = df[df["Marks"].isna()]

print("Students with missing marks:")
print(missing_marks[["Student_ID", "Student_Name", "Marks"]])
````

To distinguish absent students from marks that were simply not entered, it is better to keep an Absent indicator before converting Marks to numeric:

````python
df["Absent"] = df["Marks"].astype("string").str.strip().str.lower() == "absent"

print(df[["Student_ID", "Student_Name", "Marks", "Absent"]])
````

Risk: Deleting missing values without investigating why they are missing can remove important records and may lead to biased or inaccurate analysis

6 Identify marks outside the valid range of 0–100 and decide how they should be handled.

````python
# Identify marks outside the valid range of 0–100
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print("Invalid marks:")
print(invalid_marks[["Student_ID", "Student_Name", "Marks"]])
````

Handling: Investigate the 105 value before deleting or correcting it, because it may be a data-entry error or another special case.

Explanation: Marks should be between 0 and 100, so values outside this range are identified as invalid and should be investigated before making changes.

7.Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.

````python
# Display the cleaned DataFrame
print("Cleaned DataFrame:")
print(df)

# Report the results
print("Missing marks remaining:", df["Marks"].isna().sum())
print("Invalid marks identified:", len(invalid_marks))
````
